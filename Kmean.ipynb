```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
iris = load_iris()
y = iris.target
X = iris.data
```

```python
Kmean_random = KMeans(n_clusters = 3,random_state = 42)
Kmean_random.fit(X) #unsupervised
```

```python
y_kmean = Kmean_random.predict(X)
```

```python
sns.set(style = 'whitegrid')
plt.figure(figsize = (10,8))
plt.scatter(X[:,0],X[:,1], c = Kmean_random.labels_, cmap = 'viridis')
center = Kmean_random.cluster_centers_
plt.scatter(center[:,0],center[:,1], c = 'red', s= 200,marker = 'x')
plt.title("Without random_state (changes every run)")
plt.title('K-Means Clustering on Iris Dataset')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```
